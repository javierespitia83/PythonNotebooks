## 2.1 Branching with if

`if condition:` runs its block only when `condition` is `True`. `if/else` covers both cases. `if/elif/else` checks multiple conditions in order — the first `True` one wins, the rest are skipped. Conditionals can nest (an `if` inside another `if`), and you can chain multiple independent `if` statements (each evaluated separately, unlike `elif` which is mutually exclusive with its siblings).


```python
score = 72

if score >= 90:
    grade = 'A'
elif score >= 80:
    grade = 'B'
elif score >= 70:
    grade = 'C'
else:
    grade = 'F'
print('Grade:', grade)

# Nested conditionals
age = 20
has_id = True
if age >= 18:
    if has_id:
        print('Entry allowed')
    else:
        print('Need ID')
else:
    print('Too young')
```

## 2.2 Loops: while, for, range(), in, pass, break/continue, and the loop else

`while condition:` repeats as long as `condition` is `True`. `for item in iterable:` repeats once per item — commonly paired with `range(start, stop, step)` to loop a fixed number of times, or directly over a list/string/`in` a collection. `pass` is a no-op placeholder statement, used when Python's syntax requires a block but you have nothing to put there yet. `break` exits the loop immediately; `continue` skips to the next iteration.

**The obscure-but-real loop `else`:** both `while` and `for` support an `else` clause that runs *only if the loop finished normally* (never hit a `break`). It's built for exactly one pattern — "search for something, and do X only if not found" — and is a favorite PCEP trick question because most languages don't have it at all.


```python
# while loop
count = 0
while count < 3:
    print('while iteration', count)
    count += 1

# for loop with range()
for i in range(2, 10, 3):    # start=2, stop before 10, step 3
    print(i, end=' ')
print()

# for ... in over a collection directly
for fruit in ['apple', 'banana']:
    print(fruit, end=' ')
print()

# pass: syntactically required block, intentionally empty for now
for i in range(3):
    pass   # TODO: fill in later -- but valid, runnable code right now

# break vs continue
for i in range(10):
    if i == 5:
        break        # stop the loop entirely
    if i % 2 == 0:
        continue     # skip the rest of THIS iteration only
    print('odd, not 5+:', i)

# for-else: the else runs ONLY if the loop never hit `break`
def has_factor(n, limit):
    for i in range(2, limit):
        if n % i == 0:
            print(f'{n} is divisible by {i}')
            break
    else:
        print(f'{n} has no factor below {limit} -- looks prime-ish')

has_factor(15, 10)   # breaks early -> else is SKIPPED
has_factor(13, 10)   # loop completes without break -> else RUNS
```
