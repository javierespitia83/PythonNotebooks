## 3.1 Understand machine representation of characters

Computers store text as numbers. **ASCII** assigns numbers 0–127 to English letters, digits, and punctuation. **Unicode** extends this to essentially every character in every writing system, assigning each one a *code point* (a number, conventionally written `U+0041` for 'A' = 65). **UTF-8** is an *encoding* — a way to represent those code point numbers as bytes on disk/network; it's backward-compatible with ASCII (the first 128 code points take exactly 1 byte) and is the de facto standard.

Escape sequences let you embed special characters inside a string literal: `\n` newline, `\t` tab, `\\` literal backslash, `\'` / `\"` literal quotes, `\uXXXX` an arbitrary Unicode code point by its hex value.


```python
print('A'.encode('utf-8'))          # b'A'      -- 1 byte, same as ASCII
print('café'.encode('utf-8'))       # the é takes 2 bytes in UTF-8
print(len('café'), len('café'.encode('utf-8')))   # 4 characters, but 5 bytes

print('Line1\nLine2\tTabbed')       # \n newline, \t tab
print('Quote: \'single\' and \"double\"')
print('\u00e9')                      # 'é' via its Unicode code point in hex
```

## 3.2 Operate on strings

`ord(c)` converts a single character to its code point (integer); `chr(n)` does the reverse. Strings are **immutable** — indexing (`s[i]`) and slicing (`s[start:stop:step]`) always produce a *new* string; you can never assign to `s[i]`. Strings support `+` (concatenation), `*` (repetition), comparisons (`<`, `==`, ...) which compare code points character by character (lexicographic order), iteration with `for`, and membership with `in` / `not in`.

**Exam tip:** `'Z' < 'a'` is `True` — uppercase letters have *smaller* code points than lowercase letters in ASCII/Unicode (`'Z'` is 90, `'a'` is 97), which trips people up expecting alphabetical order.


```python
print(ord('A'), ord('a'), chr(65), chr(97))

s = 'Automate'
print(s[0], s[-1], s[2:5], s[::-1])   # index, negative index, slice, reversed via step -1
try:
    s[0] = 'a'
except TypeError as e:
    print('Strings are immutable:', e)

print('Py' + 'thon', 'ab' * 3)        # concatenation, repetition
print('Z' < 'a')                       # True! uppercase code points are SMALLER than lowercase
print('apple' < 'banana')              # normal lexicographic comparison

for ch in 'abc':
    print(ch, end=' ')
print()
print('cat' in 'concatenate', 'dog' not in 'concatenate')
```

## 3.3 Employ built-in string methods

The `isX()` family (`isalpha`, `isdigit`, `isupper`, `islower`, `isspace`, ...) check a string's content and return a bool. `.join(iterable)` glues a list of strings together using the string it's called on as separator (note the direction: `sep.join(list)`, not `list.join(sep)`). `.split(sep)` does the reverse, breaking a string into a list. `.strip()` / `.lstrip()` / `.rstrip()` remove leading/trailing whitespace (or given characters). `.find(sub)` returns the index of the first match or `-1` if not found; `.index(sub)` does the same but *raises `ValueError`* instead of returning -1; `.rfind(sub)` searches from the right. `sorted(string)` returns a sorted **list of characters** (not a string) — code-point order, so digits < uppercase < lowercase.


```python
print('Automate2026'.isalnum(), 'Automate'.isalpha(), '2026'.isdigit())

print(', '.join(['cats', 'dogs', 'mice']))   # separator.join(list)
print('a,b,,c'.split(','))                    # keeps empty strings between consecutive separators
print('   padded   '.strip())

s = 'abcabc'
print(s.find('c'), s.find('z'))       # 2, -1 (never raises)
try:
    s.index('z')                       # raises ValueError instead of returning -1
except ValueError as e:
    print('index() raised:', e)
print(s.rfind('a'))                    # search from the right: last 'a' is at index 3

print(sorted('dBcA'))                  # ['A', 'B', 'c', 'd'] -- uppercase sorts before lowercase
```

## 3.4 List methods, alongside string methods

The official PCAP prep course (Python Essentials 2, Module 2) teaches **"Strings, String and List Methods, Exceptions"** together, not strings in isolation — because both are *sequences* and share a lot of behavior. It's worth knowing exactly where they overlap and where they diverge.

**Shared sequence protocol** (works the same way on both): `len()`, indexing `s[i]`, slicing `s[a:b]`, `in` / `not in`, iteration with `for`, and even some methods with identical names and meaning — `.index(x)` (position of first match, raises `ValueError` if absent) and `.count(x)` (how many times `x` appears).

**Where they diverge — mutability:** strings are immutable, so every "modifying" string method (`.upper()`, `.strip()`, `.replace()`, ...) *returns a new string* and leaves the original untouched. Lists are mutable, so `.append()`, `.insert()`, `.remove()`, `.sort()`, `.reverse()`, `.clear()` change the list **in place** and return `None` — there's no list equivalent of `s.upper()` that returns a modified copy by convention; you mutate directly instead.

**Exam tip:** `sorted(x)` (the *function*) works on both strings and lists and always returns a **new list** without touching `x`. `x.sort()` (the *method*) only exists on lists, sorts in place, and returns `None`.

```python
text = 'banana'
items = ['b', 'a', 'n', 'a', 'n', 'a']

# Shared: len(), indexing, in, .index(), .count() behave identically
print(len(text), len(items))
print(text[1], items[1])
print('a' in text, 'a' in items)
print(text.index('n'), items.index('n'))   # first match -> 2, in both
print(text.count('a'), items.count('a'))    # 3, in both

# Diverge: string methods return NEW strings; list methods mutate IN PLACE
upper_text = text.upper()
print(text, '->', upper_text)               # original text is untouched

items.append('!')                            # mutates items directly, returns None
print(items)

# sorted() (function) vs .sort() (method)
print(sorted(text))          # works on a string too -> returns a NEW LIST of characters
print(sorted(items))         # returns a new sorted list, `items` itself is unchanged so far
items.sort()                 # mutates `items` in place, returns None
print(items)
```
