## 19.1 time.time() and time.sleep()

`time.time()` returns the current Unix timestamp (seconds since Jan 1, 1970) — handy for measuring how long code takes to run. `time.sleep(seconds)` pauses execution, useful for polling loops or rate-limiting.


```python
import time

start = time.time()
total = sum(range(1_000_000))
end = time.time()
print(f'That took {end - start:.4f} seconds. Sum = {total}')

time.sleep(0.2)
print('Slept for 0.2 seconds.')
```

## 19.2 The datetime module

`datetime.datetime` represents a calendar date and time together, supports arithmetic with `timedelta`, and `.strftime()`/`.strptime()` convert to/from formatted strings.


```python
from datetime import datetime, timedelta

now = datetime.now()
print(now)
print(now.strftime('%Y-%m-%d %H:%M'))

in_a_week = now + timedelta(days=7)
print(in_a_week.strftime('%A, %B %d'))

parsed = datetime.strptime('2026-01-15', '%Y-%m-%d')
print(parsed)
```

## 19.4 Launching other programs with subprocess

`subprocess.run(['command', 'arg1'])` launches another program and waits for it to finish, letting you capture its output — the standard way to call external tools (like `git`, `ffmpeg`, or another script) from Python.


```python
import subprocess

result = subprocess.run(['python3', '-c', 'print(2 + 2)'], capture_output=True, text=True)
print('stdout:', result.stdout.strip())
print('return code:', result.returncode)
```
