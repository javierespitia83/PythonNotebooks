## 22.4 Producing searchable-text PDFs with NAPS2

`pytesseract` (used above) extracts text *out* of an image into a Python string. Sometimes what you actually want is the opposite direction: take a scanned document and produce a new PDF where the original scanned image is kept but an invisible, searchable text layer is added underneath it — so the PDF looks identical but you can now Ctrl+F to search it. The book automates this with the free **NAPS2** scanning application's command-line interface, invoked from Python via `subprocess`: `naps2.console --ocrlang eng -o output.pdf -i scanned_page.png` scans/converts and OCRs in one step, producing that searchable PDF.

**Environment note:** NAPS2 is a desktop scanning application (Windows-first, with a Linux console build) that isn't installed in this sandbox and can't be installed without a full graphical/scanning-stack setup. The command below is the real, correct invocation from the book — run it on a machine with NAPS2 installed.

```python
import subprocess, shutil

naps2_demo_command = [
    'naps2.console',
    '--ocrlang', 'eng',
    '-o', 'searchable_output.pdf',
    '-i', 'scanned_page.png',
]

if shutil.which('naps2.console'):
    result = subprocess.run(naps2_demo_command, capture_output=True, text=True)
    print('naps2.console exit code:', result.returncode)
else:
    print('NAPS2 is not installed in this sandbox -- illustrative command from the book:')
    print(' ', ' '.join(naps2_demo_command))
    print('subprocess.run([...]) is the exact mechanism: build the argument list, then run it.')
```
