## 7.1 Creating dictionaries and accessing values

A dictionary maps *keys* to *values* using `{ }`. Unlike lists, dictionaries are unordered and indexed by (mostly) any immutable type, not just integers. Accessing a missing key raises `KeyError`.


```python
my_cat = {'size': 'fat', 'color': 'gray', 'age': 17}
print(my_cat['color'])
print('age' in my_cat)
try:
    print(my_cat['breed'])
except KeyError as e:
    print('Missing key:', e)
```

## 7.2 keys(), values(), items(), and looping

`.keys()`, `.values()`, and `.items()` return list-like views you can loop over. `.items()` gives `(key, value)` tuples, handy with the multiple-assignment trick in a `for` loop.


```python
spam = {'color': 'red', 'age': 42}
for k, v in spam.items():
    print(f'Key: {k}  Value: {v}')
```

## 7.3 get() and setdefault()

`.get(key, default)` avoids a `KeyError` by returning a fallback value if the key is missing. `.setdefault(key, default)` sets the key to `default` *only if it doesn't already exist*, and is handy for counting or building up dictionaries in a loop.


```python
picnic_items = {'apples': 5, 'cups': 2}
print(picnic_items.get('eggs', 0))

message = 'It was a bright cold day in April'
count = {}
for character in message:
    count.setdefault(character, 0)
    count[character] += 1
print(count)
```

## 7.4 Nested data structures

Dictionaries and lists can contain each other, letting you model real-world data: a list of dictionaries (rows of records), or a dictionary of dictionaries (records keyed by ID). `pprint.pprint()` formats nested structures far more readably than plain `print()`.


```python
from pprint import pprint

all_guests = {
    'Alice': {'apples': 5, 'pretzels': 12},
    'Bob': {'ham sandwiches': 3, 'apples': 2},
}
pprint(all_guests)

def total_brought(guests, item):
    return sum(v.get(item, 0) for v in guests.values())

print('Total apples:', total_brought(all_guests, 'apples'))
```
