## 8.1 String literals: quotes, escapes, raw, multiline

Strings can use single or double quotes (double quotes let you include an unescaped `'`). Escape sequences like `\n` and `\t` insert special characters. A raw string (`r'...'`) ignores escape sequences, useful for Windows paths and regular expressions. Triple-quoted strings can span multiple lines.


```python
print("That is Alice's cat.")
print('Line one\nLine two\tTabbed')
print(r'C:\Users\Alice\Desktop')   # raw string: backslashes are literal
print('''Dear Alice,

See you soon.
-Bob''')
```

## 8.2 Indexing, slicing, and membership

Strings behave like lists of characters for indexing, slicing, `len()`, and `in` / `not in` — but unlike lists, strings are immutable, so you can't assign to `s[0]`.


```python
greeting = 'Hello, world!'
print(greeting[0], greeting[-1])
print(greeting[7:-1])
print('world' in greeting)
```

## 8.3 f-strings

F-strings (`f'...'`) let you embed variables and expressions directly inside `{ }` in a string, and are the modern, most readable way to build strings from data — preferred over `+` concatenation or the older `%s` / `.format()` styles.


```python
name = 'Al'
age = 4000
print(f'My name is {name}. I am {age} years old.')
print(f'In ten years I will be {age + 10}.')
```

## 8.4 Useful string methods

`upper()`/`lower()` change case; `isX()` methods (`isalpha`, `isdecimal`, `isspace`, ...) validate content; `startswith()`/`endswith()` check the ends; `join()`/`split()` convert between a string and a list of strings; `strip()`/`lstrip()`/`rstrip()` remove whitespace (or other characters) from the ends.


```python
s = '   Hello, World!   '
print(s.strip())
print('42'.isdecimal(), 'forty two'.isdecimal())
words = 'My name is Simon'.split()
print(words)
print('-'.join(words))
print('Hello'.upper().lower())
```

## 8.5 ord(), chr(), and Unicode

Every character maps to a numeric Unicode *code point*. `ord()` converts a one-character string to its code point; `chr()` does the reverse. This is how computers ultimately represent text as numbers (most commonly encoded as UTF-8).


```python
print(ord('A'), ord('a'), ord('!'))
print(chr(65), chr(97))
print(chr(ord('A') + 1))   # 'B'
```

## 8.6 Copying and pasting strings with pyperclip

The third-party `pyperclip` module reads and writes your OS clipboard: `pyperclip.copy(text)` puts a string on the clipboard, `pyperclip.paste()` returns whatever string is currently on it. The book builds a real project around this ("Add Bullets to Wiki Markup"): read whatever text the user just copied, transform it line by line, then copy the result back so the user can paste it somewhere else — a clipboard becomes a tiny two-way pipe between your program and any other application.

**Environment note:** on Linux, `pyperclip` needs an external helper installed (`xclip`, `xsel`, or `wl-clipboard`) to actually reach the clipboard; on Windows/macOS it works out of the box. This sandbox has no clipboard mechanism at all (it's headless), so the cell below catches that specific failure — the *text-transformation logic* (the actual point of the project) is still real and verified; only the OS clipboard I/O is unavailable here.

```python
import pyperclip

def add_bullets(text):
    """The book's project: prefix each line with '* ' for wiki markup."""
    lines = text.split('\n')
    return '\n'.join('* ' + line for line in lines)

sample = "tofu\ngreen peppers\nsoy sauce"
print(add_bullets(sample))   # the transformation logic itself -- always verifiable, no clipboard needed

try:
    pyperclip.copy(sample)
    clipboard_text = pyperclip.paste()
    pyperclip.copy(add_bullets(clipboard_text))
    print('Clipboard round-trip worked -- bulleted text is now on the clipboard.')
except pyperclip.PyperclipException as e:
    print('No clipboard mechanism in this sandbox (expected, headless environment):')
    print(' ', e.args[0].splitlines()[0])
    print('On your own machine (Cursor/desktop), pyperclip.copy()/.paste() work directly.')
```
