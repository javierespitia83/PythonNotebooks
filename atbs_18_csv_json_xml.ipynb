## 18.1 Reading and writing CSV with csv.reader/writer

The `csv` module handles the fiddly bits of comma-separated files for you (quoting, embedded commas). `csv.reader`/`csv.writer` work with plain lists of values per row.


```python
import csv

with open('demo.csv', 'w', newline='') as f:
    writer = csv.writer(f)
    writer.writerow(['name', 'age'])
    writer.writerow(['Alice', 30])
    writer.writerow(['Bob', 25])

with open('demo.csv', 'r', newline='') as f:
    reader = csv.reader(f)
    for row in reader:
        print(row)
```

## 18.2 DictReader and DictWriter

`csv.DictReader` turns each row into a dictionary keyed by the header row's column names — usually more readable than indexing rows by position. `DictWriter` is the write-side counterpart.


```python
import csv

with open('demo.csv', 'r', newline='') as f:
    reader = csv.DictReader(f)
    for row in reader:
        print(row['name'], 'is', row['age'])

import os
os.remove('demo.csv')
```

## 18.3 JSON: json.dumps/loads and dump/load

`json.dumps(obj)` / `json.loads(text)` convert between a Python object and a JSON string in memory; `json.dump(obj, file)` / `json.load(file)` do the same directly to/from a file. JSON maps cleanly onto Python dicts and lists, which is why it's the most common data-interchange format for APIs.


```python
import json

data = {'name': 'Alice', 'pets': ['cat', 'dog'], 'active': True}
text = json.dumps(data, indent=2)
print(text)

parsed = json.loads(text)
print(parsed['pets'])
```

## 18.4 A brief note on XML

XML is older and more verbose than JSON but still common in enterprise/legacy systems. `xml.etree.ElementTree` parses XML into a tree you can search with `.find()` / `.findall()` and simple XPath-like paths.


```python
import xml.etree.ElementTree as ET

xml_text = '<people><person name="Alice" age="30"/><person name="Bob" age="25"/></people>'
root = ET.fromstring(xml_text)
for person in root.findall('person'):
    print(person.get('name'), person.get('age'))
```
