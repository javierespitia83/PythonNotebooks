## 6.1 Creating, indexing, and slicing lists

A list is an ordered, mutable sequence of values written between `[ ]`. Indexing with `spam[i]` gets a single item (0-based, negative indexes count from the end). Slicing with `spam[start:stop]` gets a new list containing items from `start` up to (not including) `stop`.


```python
spam = ['cat', 'bat', 'rat', 'elephant']
print(spam[0], spam[-1])
print(spam[1:3])
print(spam[:2], spam[2:])
print(len(spam))
```

## 6.2 Common list methods

Lists have methods for adding (`append`, `insert`), removing (`remove`, `del`), finding (`index`), and reordering (`sort`, `reverse`) items. Most of these mutate the list *in place* and return `None` — don't write `spam = spam.append(x)`.


```python
spam = ['cat', 'dog', 'bat']
spam.append('moose')
spam.insert(1, 'chicken')
print(spam)
spam.remove('bat')
print(spam)
print(spam.index('moose'))
spam.sort()
print(spam)
```

## 6.3 Looping over lists: for, enumerate(), in

A `for` loop iterates directly over a list's items. `enumerate()` gives you both the index and the item when you need both. The `in` / `not in` operators check membership and return a Boolean.


```python
supplies = ['pens', 'staplers', 'flamethrowers', 'binders']
for index, item in enumerate(supplies):
    print(f'Index {index}: {item}')

print('cat' in supplies, 'pens' in supplies)
```

## 6.4 Tuples and mutability

Tuples look like lists but use `( )` and are *immutable*: once created, you can't change, add, or remove items. Use a tuple when a sequence of values should never change; Python can also optimize tuple code slightly better than list code.


```python
eggs = ('hello', 42, 0.5)
print(eggs[0], len(eggs))
try:
    eggs[1] = 99
except TypeError as e:
    print('Error:', e)

print(list(eggs))     # convert tuple -> list
print(tuple([1, 2]))  # convert list -> tuple
```

## 6.5 References: why copying a list is different from copying a number

Variables holding lists store a *reference* to the list, not the list itself. Assigning `eggs = spam` makes both names point to the *same* list, so mutating one affects the other. Use `copy.copy()` (or `copy.deepcopy()` for nested lists) when you need an independent duplicate.


```python
import copy

spam = [1, 2, 3]
eggs = spam            # eggs and spam refer to the SAME list
eggs.append(4)
print(spam)            # also changed!

cheese = copy.copy(spam)  # an independent copy
cheese.append(99)
print(spam, cheese)
```
