## 3.1 Lists: vectors, indexing, slicing, methods, comprehensions, copying

A list is an ordered, mutable sequence (`[ ]`). Index with `lst[i]` (0-based, negative from the end), slice with `lst[a:b:step]`. `len()` gives the item count. Key mutating methods: `.append()`, `.insert()`, `.remove()`, `.pop()`, `.sort()`, `.reverse()`, `.extend()`, `.clear()`; `.index()` and `sorted()` (the function, returns a new list) don't mutate. `del lst[i]` removes by index. A **list comprehension** `[expr for item in iterable if cond]` builds a list in one line.

**Copying vs. cloning:** `b = a` does *not* copy a list — it makes `b` another name for the *same* list, so mutating one mutates both. To get an independent copy, use `a.copy()`, `list(a)`, or a full slice `a[:]`.


```python
vec = [10, 20, 30, 40, 50]
print(vec[1], vec[-1], vec[1:3], len(vec))

vec.append(60)
vec.insert(0, 0)
print(vec)
del vec[1]                      # remove by index
print(vec)

squares = [x**2 for x in range(6)]                  # list comprehension
even_squares = [x**2 for x in range(6) if x % 2 == 0]
print(squares, even_squares)

matrix = [[1, 2, 3], [4, 5, 6]]     # list of lists -- a "matrix"
print(matrix[1][2])                  # row 1, column 2 -> 6

# Copying vs. cloning -- a classic PCEP trap
a = [1, 2, 3]
b = a                # NOT a copy -- same list, two names
b.append(4)
print('a:', a, ' b:', b)          # a changed too!

c_clone = a.copy()    # an independent copy
c_clone.append(99)
print('a:', a, ' clone:', c_clone)   # a is untouched this time
```

## 3.2 Tuples: immutable sequences

Tuples (`( )`) support indexing and slicing just like lists but are immutable. A tuple can be nested inside a list, and vice versa. Prefer tuples for fixed, related values that should never change (a coordinate pair, a database row).


```python
point = (3, 4)
print(point[0], point[-1], point[:1])

try:
    point[0] = 99
except TypeError as e:
    print('Immutable:', e)

list_of_tuples = [(1, 'a'), (2, 'b')]      # tuples inside a list
list_of_tuples.append((3, 'c'))
tuple_of_lists = ([1, 2], [3, 4])           # lists inside a tuple -- the inner LISTS are still mutable
tuple_of_lists[0].append(99)
print(list_of_tuples)
print(tuple_of_lists)
```

## 3.3 Dictionaries: keys, values, iterating

A dictionary maps keys to values (`{key: value}`), built and modified with `d[key] = value`, and read with `d[key]` (raises `KeyError` if missing). Check for a key with `key in d` before that, to avoid the error. `.keys()`, `.values()`, `.items()` give you views for looping.


```python
stock = {'apples': 10, 'bananas': 5}
stock['cherries'] = 20            # add a new key
stock['apples'] += 3               # update an existing key
print(stock)
del stock['bananas']               # remove a key
print(stock)

print('cherries' in stock, 'bananas' in stock)

for key, value in stock.items():
    print(f'{key}: {value}')

print(list(stock.keys()), list(stock.values()))
```

## 3.4 Strings, briefly (see 03_strings.ipynb for the full PCAP-level deep dive)

Strings are immutable sequences (`'...'` or `"..."`). Escape special characters with `\` (e.g. `\n`, `\'`), or avoid escaping altogether by picking the other quote style (`"it's fine"` needs no escape). Triple quotes (`'''...'''`) make multi-line strings. Basic functions/methods: `len()`, `.upper()`, `.lower()`, `.strip()`, string slicing works exactly like list slicing.


```python
s = "It's a nice day"        # double quotes avoid escaping the apostrophe
s2 = 'He said \'hi\''         # or escape it with a backslash inside single quotes

multiline = '''Line one
Line two'''
print(s, '|', s2)
print(multiline)

print(len(s), s[:4], s.upper())
```
