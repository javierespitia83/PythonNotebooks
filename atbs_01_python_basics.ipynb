## 1.1 Expressions and math operators

An *expression* is a piece of code that reduces to a single value: a value on its own (like `2`), or values combined with operators (like `2 + 2`). Python's math operators follow the usual order of operations (`**` first, then `* / // %`, then `+ -`).


```python
print(2 + 3 * 6)
print((2 + 3) * 6)
print(2 ** 8)      # exponentiation
print(23 // 7)     # integer (floor) division
print(23 % 7)      # modulus / remainder
print(23 / 7)      # regular division always returns a float
```

## 1.2 The int, float, and string data types

Every value has a data type. Whole numbers are `int`, decimal numbers are `float`, and text wrapped in quotes is `str`. Mixing an int and a float in an operation always produces a float. The `+` and `*` operators change meaning for strings: `+` concatenates, `*` repeats.


```python
print(type(42), type(42.0), type('42'))
print(3 + 4.0)          # int + float -> float
print('Alice' + 'Bob')  # string concatenation
print('Alice' * 3)      # string replication
# print('Alice' + 42)   # TypeError: can only concatenate str (not "int") to str
```

## 1.3 Variables and assignment

A variable is a name attached to a value (think name tag, not box — this matters later for mutable types like lists). Assignment statements use `=`. Reassigning a variable simply moves the name tag to a new value; the old value is forgotten if nothing else refers to it.


```python
spam = 40
eggs = 2
print(spam + eggs)
spam = spam + 2   # overwrite spam with a new value
print(spam)

# Valid names: letters, numbers, underscore; can't start with a number or be a keyword.
current_balance = 100.50
print(current_balance)
```

## 1.4 print(), input(), and your first program

`print()` displays text on the screen. `input()` pauses the program, waits for the user to type something and press Enter, and always returns a string — even if the user typed a number.


```python
print('Hello, world!')
name = input('What is your name? ')
print('It is good to meet you, ' + name)
print('The length of your name is:', len(name))
```

## 1.5 Useful built-in functions: len(), str(), int(), float(), type(), round(), abs()

`len()` counts characters in a string (or items in a list). `str()`, `int()`, and `float()` convert between data types — you'll need `int()` a lot, since `input()` always returns a string. `type()` tells you a value's data type. `round()` rounds to the nearest integer (or to N decimal places), and `abs()` returns the absolute value.


```python
age_text = '29'
age = int(age_text)
print('I am ' + str(age + 1) + ' next year.')
print(type(age), type(age_text))
print(round(3.14159, 2))
print(round(2.5), round(3.5))  # banker's rounding: rounds to nearest even
print(abs(-25))
```
