## 23.4 Safety: the fail-safe

Because GUI automation controls your *actual* mouse and keyboard, a runaway script can be hard to stop. PyAutoGUI's fail-safe (on by default) lets you abort instantly by slamming the mouse to a screen corner — never disable it while developing a new script.


```python
# import pyautogui
# pyautogui.FAILSAFE  # True by default: moving the mouse to a corner raises FailSafeException
print('Keep FAILSAFE on. It is your emergency stop button.')
```

## 23.5 Image recognition on screen

Instead of hardcoding pixel coordinates (fragile — they break the moment a window moves), `pyautogui.locateOnScreen('button.png')` searches the current screen for a region matching a reference image you saved earlier, and returns its location (or `None` if not found) — so your script can find "the Submit button" wherever it currently is. `locateAllOnScreen()` returns every match instead of just the first. `pixelMatchesColor(x, y, (r, g, b))` checks whether a specific pixel is close to a given color, useful for waiting until something changes on screen (like a progress bar finishing).

**Environment note:** this needs an actual screen to look at — this headless sandbox has none, so these calls raise immediately.

```python
try:
    import pyautogui
    location = pyautogui.locateOnScreen('submit_button.png')
    print('Found at:', location)
except Exception as e:
    print(f'{type(e).__name__} in this sandbox (expected, no real screen/display available here):', e)
    print('On your own desktop, this returns the (left, top, width, height) box where the image was found.')
```

## 23.6 Finding and controlling windows

`PyAutoGUI` (via the bundled `pygetwindow` package) can also find and manipulate actual application windows: `getActiveWindow()` returns the currently focused window, `getWindowsWithTitle('Notepad')` searches by (partial) title, and the returned window object has `.moveTo(x, y)`, `.resizeTo(width, height)`, `.maximize()`, `.minimize()`, and `.close()`. This lets a script arrange multiple application windows automatically, or bring a specific one to the front before typing into it.

**Environment note:** again, this needs a real windowing desktop session, which this headless sandbox doesn't have.

```python
try:
    import pyautogui
    active = pyautogui.getActiveWindow()
    print('Active window title:', active.title if active else None)
except Exception as e:
    print(f'{type(e).__name__} in this sandbox (expected, no desktop window manager/display here):', e)
    print('On your own desktop: active.moveTo(0, 0), active.resizeTo(800, 600), active.maximize(), etc.')
```
