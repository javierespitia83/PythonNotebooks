## 10.1 File paths with pathlib

The `pathlib` module represents file paths as objects instead of raw strings, so you can join paths with `/` and it handles Windows vs. macOS/Linux separators for you. `Path.cwd()` gets the current working directory.


```python
from pathlib import Path

p = Path.cwd() / 'demo_folder' / 'notes.txt'
print(p)
print(p.name, p.suffix, p.parent)
```

## 10.2 Writing and reading text files

`open(filename, mode)` returns a file object. `'w'` mode creates/overwrites a file for writing, `'a'` appends, `'r'` (the default) reads. `with open(...) as f:` automatically closes the file for you when the block ends — always prefer it over manual `open()`/`close()`.


```python
from pathlib import Path

path = Path('demo.txt')
with open(path, 'w') as f:
    f.write('Hello, world!\n')
    f.write('Second line.\n')

with open(path, 'r') as f:
    content = f.read()
print(content)

with open(path, 'r') as f:
    for line in f.readlines():
        print(repr(line))

path.unlink()  # clean up the demo file
```

## 10.3 Appending to a file

Opening in `'a'` mode adds new content to the end of an existing file instead of erasing it — useful for log files that grow over time.


```python
from pathlib import Path

path = Path('log_demo.txt')
with open(path, 'w') as f:
    f.write('First run\n')
with open(path, 'a') as f:
    f.write('Second run\n')

print(path.read_text())
path.unlink()
```

## 10.4 Saving structured data: json and shelve

Plain text files only store strings. To save Python data structures (lists, dicts) directly, use the `json` module for a human-readable, portable format, or the `shelve` module for a quick binary key-value store local to your machine.


```python
import json
from pathlib import Path

data = {'name': 'Zophie', 'age': 8, 'likes': ['naps', 'treats']}
path = Path('cat.json')
path.write_text(json.dumps(data))

loaded = json.loads(path.read_text())
print(loaded)
path.unlink()

import shelve
with shelve.open('demo_shelf') as db:
    db['cats'] = ['Zophie', 'Pooka']
with shelve.open('demo_shelf') as db:
    print(list(db.keys()), db['cats'])
```
