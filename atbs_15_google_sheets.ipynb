## 15.1 Why EZSheets instead of the raw Google API

Google Sheets has its own official API, but it's verbose to use directly. The third-party `ezsheets` package wraps it with a much simpler interface, at the cost of needing an initial OAuth authorization (a `credentials-sheets.json` file from Google Cloud Console) the first time you run it.
