## 2.1 Booleans and comparison operators

The Boolean type has exactly two values, `True` and `False` (capitalized, no quotes). Comparison operators (`== != < > <= >=`) compare two values and always evaluate to a Boolean.


```python
print(42 == 42, 42 == 99)
print('hello' == 'Hello')   # case-sensitive
print(42 == 42.0)          # int/float compare by value
print(42 == '42')          # int is never equal to the string form
age = 29
print(age >= 18)
```

## 2.2 Boolean operators: and, or, not

`and` is True only if both sides are True. `or` is True if at least one side is True. `not` flips a single Boolean. These combine naturally with comparison operators to build more complex conditions.


```python
temp = 15
is_raining = True
print(temp < 20 and is_raining)
print(temp < 20 or is_raining)
print(not is_raining)
print((4 < 5) and (9 < 6))
```

## 2.3 if, elif, and else statements

An `if` clause runs only when its condition is `True`. `elif` ("else if") checks another condition, but only if all previous conditions were `False`. `else` catches everything else. Order matters: once a True condition is found, the rest are skipped — so put more specific conditions before more general ones.


```python
name = 'Carol'
age = 3000

if name == 'Alice':
    print('Hi, Alice.')
elif age < 12:
    print('You are not Alice, kiddo.')
elif age > 2000:
    print('Unlike you, Alice is not an undead, immortal vampire.')
elif age > 100:
    print('You are not Alice, grannie.')
else:
    print('You are neither Alice nor a little kid.')
```

## 2.4 Blocks and indentation

Python groups lines of code into *blocks* using indentation instead of curly braces. A block starts when a line ends with `:` and the next line is indented further; it ends when the code returns to the previous indentation level. Blocks can be nested inside other blocks.


```python
username = 'Mary'
password = 'swordfish'

if username == 'Mary':
    print('Hello, Mary')
    if password == 'swordfish':
        print('Access granted.')
    else:
        print('Wrong password.')
```
