## 5.1 Raising your own exceptions

`raise Exception('message')` deliberately stops execution and hands control to the nearest matching `except`. It's common to `raise` inside a function (because the function doesn't know how the caller wants to handle the problem) and `try`/`except` where the function is called.


```python
def box_print(symbol, width, height):
    if len(symbol) != 1:
        raise Exception('Symbol must be a single character string.')
    if width <= 2 or height <= 2:
        raise Exception('Width and height must be greater than 2.')
    print(symbol * width)
    for i in range(height - 2):
        print(symbol + (' ' * (width - 2)) + symbol)
    print(symbol * width)

try:
    box_print('*', 4, 4)
    box_print('ZZ', 3, 3)   # invalid: symbol too long
except Exception as err:
    print('An exception happened:', err)
```

## 5.2 assert statements

`assert condition, message` is a sanity check for the *programmer*, not the user: if the condition is False it raises `AssertionError` and the program should crash, because it means your own logic is broken. Never catch an AssertionError with try/except — let it fail loudly and fast.


```python
ages = [26, 57, 92, 54, 22, 15, 17, 80, 47, 73]
ages.sort()
assert ages[0] <= ages[-1]   # sanity check: sort() should keep this true
print('Sorted:', ages)

# assert 1 == 2, 'this always fails, just to show the AssertionError shape'
```

## 5.3 The logging module

`print()` debugging clutters your program and you have to remember to remove it. `logging` gives you leveled, timestamped messages you can leave in the code and switch off with one line (`logging.disable()`). Levels, from least to most severe: DEBUG, INFO, WARNING, ERROR, CRITICAL.


```python
import logging
logging.basicConfig(level=logging.DEBUG, format='%(asctime)s - %(levelname)s - %(message)s')

def factorial(n):
    logging.debug(f'Start of factorial({n})')
    total = 1
    for i in range(1, n + 1):
        total *= i
        logging.debug(f'i is {i}, total is {total}')
    logging.debug(f'End of factorial({n})')
    return total

print(factorial(5))
# logging.disable(logging.CRITICAL)  # uncomment to silence all log messages
```

## 5.4 Debugging strategy without an IDE debugger

In an editor like Cursor you'd normally set breakpoints and step through code with a visual debugger. In a plain notebook cell, Python's built-in `pdb` module gives you the same tool: `breakpoint()` drops you into an interactive debugger at that exact line (type `n` to step, `p varname` to inspect, `c` to continue).


```python
def buggy_total(numbers):
    total = 0
    for n in numbers:
        # breakpoint()   # uncomment to step through interactively
        total += n
    return total

print(buggy_total([1, 2, 3, 4]))
```
