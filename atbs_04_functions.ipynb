## 4.1 Defining and calling functions

`def` creates a function; the code inside only runs when the function is *called*, not when it's defined. Parameters are variables that receive the arguments passed at the call site, and they only exist inside that function call (local scope — more on this below).


```python
def say_hello_to(name):
    print('Good morning, ' + name)
    print('Good evening, ' + name)

say_hello_to('Alice')
say_hello_to('Bob')
```

## 4.2 Return values and None

A function call is itself an expression: it evaluates to whatever the `return` statement specifies. If a function has no `return` statement (or a bare `return`), it implicitly returns `None`, Python's "no value" value.


```python
def get_answer(n):
    if n == 1:
        return 'It is certain'
    elif n == 2:
        return 'Ask again later'
    else:
        return 'Very doubtful'

print(get_answer(1))

def no_return():
    print('doing something')

result = no_return()
print(result)   # None
```

## 4.3 Keyword arguments

Some functions accept optional *keyword arguments* identified by name rather than position. `print()`'s `end` and `sep` parameters are common examples: `end` replaces the default trailing newline, `sep` replaces the default single-space separator between arguments.


```python
print('Hello', end='')
print('World')      # printed right after 'Hello' on the same line

print('cats', 'dogs', 'mice', sep=', ')
```

## 4.4 Local vs global scope

Variables created inside a function live in a *local scope* that's destroyed when the function returns — code outside the function can't see them. Variables created outside every function live in the *global scope*. If a function needs to modify a global variable (not just read it), it must declare `global varname` first.


```python
def spam():
    eggs = 'spam local'   # local variable, forgotten when spam() returns
    print(eggs)

spam()
# print(eggs)   # would raise NameError: eggs only existed inside spam()

def change_global():
    global counter
    counter += 1

counter = 0
change_global()
change_global()
print(counter)
```

## 4.5 A first look at try/except

When Python hits invalid code at runtime (like dividing by zero) it raises an *exception*, which crashes the program unless you catch it. Code that might fail goes in a `try` block; code that handles the failure goes in the matching `except` block.


```python
def safe_divide(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        print('Error: cannot divide by zero.')
        return None

print(safe_divide(10, 2))
print(safe_divide(10, 0))
```
