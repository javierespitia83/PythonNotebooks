## 1.1 Fundamental terms: interpreting vs. compiling

A **compiler** translates an entire program into machine code *before* it runs, producing a standalone executable. An **interpreter** reads and executes a program's instructions one at a time, translating as it goes — this is what `python` does with your `.py` files (technically CPython compiles to bytecode first and then interprets *that*, but conceptually you write code and run it directly, no separate build step).

Three linguistic terms, borrowed from human language, describe levels of correctness: **lexis** (the vocabulary — is `pint` a valid word/token?), **syntax** (the grammar rules — is `if x:` structured correctly?), and **semantics** (the meaning — does `x = x + 1` actually do what you intend?). Code can be syntactically valid but semantically wrong (it runs, but does the wrong thing).


```python
# Semantically "wrong" but syntactically perfectly valid Python:
principal = 100
interest_rate = 0.05
# Bug: multiplied instead of adding the interest -- syntax is fine, semantics are wrong
total = principal * interest_rate
print('Buggy total:', total)   # should probably be principal + principal * interest_rate
total_correct = principal + principal * interest_rate
print('Intended total:', total_correct)

# A syntax error would look like this (uncomment to see it -- it won't even run):
# if x
#     print('missing colon')
```

## 1.2 Python's logic and structure

**Keywords** are reserved words with special meaning (`if`, `for`, `def`, `class`, ...) — you can't use them as variable names. An **instruction** (statement) is one complete step of code. **Indentation** (whitespace at the start of a line) is how Python marks blocks of code — not just a style choice, it's part of the syntax (this is different from most other languages, which use `{ }`). **Comments** start with `#` and are ignored by Python entirely; they're notes for humans.


```python
import keyword
print(keyword.kwlist[:10])         # first 10 of Python's reserved keywords
print(len(keyword.kwlist), 'total keywords')
print(keyword.iskeyword('for'), keyword.iskeyword('data'))

# Indentation defines the block -- this is REQUIRED, not optional style:
temperature = 30
if temperature > 25:
    print('Hot')          # part of the if-block: indented
    print('Wear a hat')   # also part of the if-block
print('This always runs')  # NOT indented -- outside the if-block

# A comment: this line does nothing when Python runs it
# print('this is commented out and never executes')
```

## 1.3 Literals, variables, and numeral systems

A **literal** is a value written directly in code (`42`, `3.14`, `'hi'`, `True`). Python's basic literal types: `bool` (`True`/`False`), `int`, `float` (including **scientific notation**, e.g. `2.5e3` = 2500.0), and `str`.
```python
scientific = 2.5e3          # 2.5 * 10^3
print(scientific)

binary_literal = 0b1010     # binary for 10
octal_literal = 0o12        # octal for 10
hex_literal = 0x0A          # hex for 10
print(binary_literal, octal_literal, hex_literal)   # all print as decimal: 10 10 10

print(bin(10), oct(10), hex(10))   # decimal -> string in that base

# PEP-8 naming conventions
max_retry_count = 3         # snake_case for variables
DEFAULT_TIMEOUT = 30        # UPPER_CASE for constants


class HttpClient:           # CapWords for classes
    pass
```

## 1.4 Operators: numeric, string, bitwise, boolean, relational

Numeric operators: `** * / % // + -`. String operators reuse `+` (concatenation) and `*` (repetition). Assignment has *shortcut* forms (`+=`, `-=`, `*=`, ...). **Bitwise operators** work on the binary representation of integers: `&` (AND), `|` (OR), `^` (XOR), `~` (NOT, flips all bits), `<<` (shift left = multiply by 2 per shift), `>>` (shift right = integer-divide by 2 per shift). Boolean operators (`not`, `and`, `or`) combine `True`/`False` values. Relational operators (`== != > >= < <=`) compare values. Operators have **precedence** (order of evaluation) just like in math class — `**` binds tightest, then `* / // %`, then `+ -`; parentheses always override.

**Floating-point accuracy:** floats are stored in binary, and many decimal fractions (like 0.1) have no *exact* binary representation — leading to tiny rounding errors that surprise beginners. **Type casting** converts a value from one type to another with `int()`, `float()`, `str()`, `bool()`.


```python
# Bitwise operators (on the binary representation of integers)
a, b = 5, 3            # binary: 101, 011
print(a & b)            # AND -> 001 =  1
print(a | b)            # OR  -> 111 =  7
print(a ^ b)             # XOR -> 110 =  6
print(~a)                # NOT -> flips all bits -> -(a+1) = -6
print(a << 1, a >> 1)    # shift left/right -> 10, 2

# Operator precedence: ** first, then * / // %, then + -
print(2 + 3 * 2 ** 2)    # 2 + 3*4 = 14, not (2+3)*2**2

# The classic floating-point accuracy trap:
print(0.1 + 0.2)             # NOT exactly 0.3!
print(0.1 + 0.2 == 0.3)      # False -- never compare floats with == directly
print(round(0.1 + 0.2, 2) == 0.3)   # True -- round first, or use a tolerance

# Type casting
print(int('42'), float('3.14'), str(42), bool(0), bool(''), bool('a'))
```

## 1.5 Console input/output operations

`print(*values, sep=' ', end='\n')` displays text: `sep` controls the separator *between* multiple arguments (default: a space), `end` controls what's printed *after* everything (default: a newline). `input(prompt)` always returns a `str`, even if the user types a number — you must convert it yourself with `int()` or `float()` when you need to do math with it.


```python
print('cats', 'dogs', 'mice', sep=', ')      # custom separator between args
print('Loading', end='')
print('...', end='')
print(' done')                                # end='' avoided the newlines above

# input() ALWAYS returns a string
age_text = '30'      # simulating what input('Enter your age: ') would give
print(type(age_text))
age = int(age_text)   # must convert before doing math
print(age + 1, type(age))
```
