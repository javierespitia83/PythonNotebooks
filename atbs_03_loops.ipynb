## 3.1 while loops

A `while` loop re-checks its condition every time it finishes its block, and keeps looping as long as the condition is `True`. If the condition never becomes False, you get an infinite loop.


```python
spam = 0
while spam < 5:
    print('Hello, world.')
    spam = spam + 1
print('Done.')
```

## 3.2 break and continue

`break` immediately exits the innermost loop. `continue` skips the rest of the current iteration and jumps back to re-check the loop's condition. Both only work inside `while` or `for` loops.


```python
# break: stop as soon as we find what we want
while True:
    name = 'your name' if True else input('>')  # simulated input for the demo
    if name == 'your name':
        break
print('Thank you!')

# continue: skip non-matching values
for i in range(10):
    if i % 2 != 0:
        continue   # skip odd numbers
    print(i, end=' ')
print()
```

## 3.3 for loops and range()

A `for` loop repeats its block once per item in a sequence. `range(n)` produces integers from 0 up to (not including) `n`. `range(start, stop)` and `range(start, stop, step)` give you more control, including counting backward with a negative step.


```python
for i in range(5):
    print('Iteration', i)

for i in range(2, 10, 2):   # start at 2, stop before 10, step by 2
    print(i, end=' ')
print()

for i in range(5, 0, -1):   # count down
    print(i, end=' ')
print()

total = 0
for num in range(101):
    total += num
print('Sum 0..100 =', total)
```

## 3.4 Importing modules

Python's standard library ships with *modules* — extra functions you unlock with `import`. `import random, sys` imports several modules at once. `sys.exit()` ends a program early.


```python
import random
for i in range(5):
    print(random.randint(1, 10), end=' ')
print()

import sys
def check(n):
    if n < 0:
        print('Negative not allowed, stopping.')
        # In a script this would end the program; in a notebook avoid calling it directly.
        return False
    return True

print(check(5), check(-1))
```
